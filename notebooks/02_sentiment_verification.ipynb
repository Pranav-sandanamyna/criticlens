```python
import pandas as pd
from pathlib import Path

BASE_DIR = Path().resolve().parent
csv_path = BASE_DIR / "data" / "processed" / "imdb_with_sentiment.csv"

df = pd.read_csv(csv_path)
print(f"Shape: {df.shape}")
df.head(5)
```

```python
print("Sentiment Label Distribution:")
print(df["sentiment_label"].value_counts())
print(f"\nAverage compound score: {df['sentiment_compound'].mean():.4f}")
print(f"Most positive movie: {df.loc[df['sentiment_compound'].idxmax(), 'title']}")
print(f"Most negative movie: {df.loc[df['sentiment_compound'].idxmin(), 'title']}")
```

```python
key_cols = ["plot", "sentiment_compound", "metascore", "box_office"]
print("Missing values in key columns:")
print(df[key_cols].isnull().sum())
```

```python
# Should have 250 rows
assert len(df) == 250, f"Expected 250, got {len(df)}"

# Should have sentiment scores
assert df["sentiment_compound"].notna().sum() > 200, "Too many missing sentiment scores"

print("All checks passed.")
```
