```python
import pandas as pd
from pathlib import Path

# This makes the path work regardless of where the notebook is running from
BASE_DIR = Path().resolve().parent  # Goes up from notebooks/ to criticlens/
csv_path = BASE_DIR / "data" / "raw" / "imdb_top250.csv"

print(f"Looking for file at: {csv_path}")
print(f"File exists: {csv_path.exists()}")

df = pd.read_csv(csv_path)
print(f"Shape: {df.shape}")
df.head(10)
```

```python
print(df.isnull().sum())
print("\nValue counts for genre:")
print(df['genre'].value_counts().head(10))
```

```python
print(df.describe())
```

```python
# Should have 250 rows
assert len(df) == 250, f"Expected 250 rows, got {len(df)}"

# Should have all columns
expected_cols = ['title', 'year', 'rating', 'votes', 'imdb_id', 'genre']
for col in expected_cols:
    assert col in df.columns, f"Missing column: {col}"

print("All checks passed. Data looks good.")
```
